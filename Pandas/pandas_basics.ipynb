<H1>Grouping</H1>.
groupby() is used to group rows that share the same value in a column, usually to apply an aggregation like sum, mean etc.

```python
dfs.groupby('Transmission Type').MSRP.mean()
```

```python
dfs.groupby('Transmission Type').MSRP.min()
```

```python
dfs.groupby('Transmission Type').MSRP.max()
```
